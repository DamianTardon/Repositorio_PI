```python
nombre = "Juan"
edad = 30
saludo = "¡Hola!"

# Abre el archivo en modo escritura
with open("saludos.txt", "w") as archivo:
    # Escribe variables usando f-strings y añade un salto de línea (\n)
    archivo.write(f"{saludo} Mi nombre es {nombre} y tengo {edad} años.\n")

# Al salir del bloque 'with', el archivo se cierra automáticamente
print("Se ha escrito el contenido en saludos.txt")
```
