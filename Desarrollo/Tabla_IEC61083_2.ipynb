```python
import pandas as pd

# Definición de los datos de la Tabla A.1 (IEC 61083-2)
# Datos extraídos de las tablas en las páginas 12 a 17 del PDF.
data = [
    # Designation, Description, Ut(kV), T1(us), T2(us), Beta'(%)
    ("LI-A1", "Superposition of two ideal exponential functions", 1049.60, 0.840, 60.16, 0.0),
    ("LI-A2", "Slow oscillations", 1037.60, 1.693, 47.48, 5.1),
    ("LI-A3", "Fast oscillations", 1000.20, 1.117, 48.15, 4.6),
    ("LI-A4", "f = 500 kHz Overshoot 8%", 856.01, 0.841, 47.80, 7.9),
    ("LI-A5", "f = 200 kHz Overshoot 8%", 71.972, 1.711, 47.71, 7.7),
    ("LI-A6", "f = 200 kHz Overshoot 18%", 100.17, 1.762, 41.58, 17.7),
    ("LI-A7", "f = 200 kHz Overshoot 20%", 104.35, 2.122, 38.36, 20.1),
    ("LI-A8", "f = 250 kHz Overshoot 15%", 96.012, 1.503, 44.92, 14.8),
    ("LI-A9", "f = 300 kHz Overshoot 4%", 55.928, 1.215, 55.74, 4.0),
    ("LI-A10", "f = 400 kHz Overshoot 12%", 81.929, 0.924, 42.66, 12.0),
    ("LI-A11", "f = 500 kHz Overshoot 4%", 86.597, 0.578, 56.37, 4.1),
    ("LI-A12", "f = 900 kHz Overshoot 2%", 85.584, 0.587, 57.36, 2.3),
    ("LI-M1", "Front oscillations", 952.09, 1.123, 85.60, 2.1),
    ("LI-M2", "Long duration overshoot", -1041.70, 3.356, 61.25, 9.2),
    ("LI-M3", "Short duration overshoot", -1026.50, 2.150, 41.75, 9.2),
    ("LI-M4", "Oscillations, transformer testing", -267.14, 0.987, 56.22, 4.8),
    ("LI-M5", "Oscillations, transformer testing", -55.003, 2.746, 42.11, 18.7),
    ("LI-M6", "Oscillations, transformer testing", -166.87, 1.356, 54.74, 3.8),
    ("LI-M7", "Oscillations, transformer testing", -1272.30, 1.482, 50.03, 11.2),
    ("LI-M8", "Long front, smooth", -99.732, 1.515, 49.36, -0.5),
    ("LI-M9", "Short front, some overshoot", -100.04, 0.828, 46.65, 1.4),
    ("LI-M10", "Heavy front oscillations", 100.26, 1.666, 60.85, 0.0),
    ("LI-M11", "Heavy front oscillations", 299.32, 1.661, 60.95, -0.5),
    ("LI-M12", "Changing offset level, oscillations at peak", -4.3193, 1.292, 52.27, -1.8),
    ("LI-M13", "Oscillations after onset", 39.460, 1.537, 46.94, 1.8),
    ("LI-M14", "Oscillations after onset, and overshoot", 48.549, 0.933, 37.48, 4.3),
    ("LI-M15", "Oscillations after onset, and overshoot", 497.97, 1.017, 59.19, -0.1),
    ("LI-M16", "Front oscillations", 369.21, 0.920, 47.53, 0.8),
    ("LI-M17", "Heavy oscillations at front and peak", -99.346, 1.775, 53.31, 1.3)
]

# Definición de límites fijos según Tabla A.1 para grupo LI
LIMIT_UT_PCT = 0.10  # %
LIMIT_T1_PCT = 2.0   # %
LIMIT_T2_PCT = 1.0   # %
LIMIT_BETA_ABS = 1.0 # % absoluto

# --- Generación de Datos para HOJA 1 (Original) ---
hoja1_data = []
for row in data:
    hoja1_data.append({
        "Designation in TDG": row[0],
        "Description": row[1],
        "Ut Reference value kV": row[2],
        "Ut Acceptance limit % of Ut": f"± {LIMIT_UT_PCT:.2f}",
        "T1 Reference value µs": row[3],
        "T1 Acceptance limit % of T1": f"± {int(LIMIT_T1_PCT)}",
        "T2 Reference value µs": row[4],
        "T2 Acceptance limit % of T2": f"± {LIMIT_T2_PCT:.1f}",
        "Beta' Reference value %": row[5],
        "Beta' Acceptance limit % abs": f"± {LIMIT_BETA_ABS:.1f}"
    })

df_hoja1 = pd.DataFrame(hoja1_data)

# --- Generación de Datos para HOJA 2 (Calculada) ---
hoja2_data = []

for row in data:
    desig = row[0]
    ut_ref = row[2]
    t1_ref = row[3]
    t2_ref = row[4]
    beta_ref = row[5]

    # Cálculos Ut (± 0.10%)
    ut_max = ut_ref * (1 + LIMIT_UT_PCT/100)
    ut_min = ut_ref * (1 - LIMIT_UT_PCT/100)

    # Cálculos T1 (± 2%)
    t1_max = t1_ref * (1 + LIMIT_T1_PCT/100)
    t1_min = t1_ref * (1 - LIMIT_T1_PCT/100)

    # Cálculos T2 (± 1.0%)
    t2_max = t2_ref * (1 + LIMIT_T2_PCT/100)
    t2_min = t2_ref * (1 - LIMIT_T2_PCT/100)

    # Cálculos Beta' (± 1.0 absoluto) - Opción A Matemática
    beta_max = beta_ref + LIMIT_BETA_ABS
    beta_min = beta_ref - LIMIT_BETA_ABS

    # Fila Superior (Máximos)
    hoja2_data.append({
        "Designation in TDG": desig,
        "Ut Reference value kV": ut_ref,
        "Ut Acceptance limit kV": round(ut_max, 3), # Redondeo razonable para visualización
        "T1 Reference value µs": t1_ref,
        "T1 Acceptance limit µs": round(t1_max, 4),
        "T2 Reference value µs": t2_ref,
        "T2 Acceptance limit µs": round(t2_max, 3),
        "Beta' Reference value %": beta_ref,
        "Beta' Acceptance limit %": round(beta_max, 2)
    })

    # Fila Inferior (Mínimos) - Dejamos referencia vacía para limpieza visual o se repite si se prefiere
    hoja2_data.append({
        "Designation in TDG": "", # Vacío para efecto visual de bloque
        "Ut Reference value kV": "",
        "Ut Acceptance limit kV": round(ut_min, 3),
        "T1 Reference value µs": "",
        "T1 Acceptance limit µs": round(t1_min, 4),
        "T2 Reference value µs": "",
        "T2 Acceptance limit µs": round(t2_min, 3),
        "Beta' Reference value %": "",
        "Beta' Acceptance limit %": round(beta_min, 2)
    })

df_hoja2 = pd.DataFrame(hoja2_data)

# Guardar en Excel
output_file = "Tabla_IEC61083-2_A1.xlsx"
with pd.ExcelWriter(output_file, engine='openpyxl') as writer:
    df_hoja1.to_excel(writer, sheet_name='HOJA 1', index=False)
    df_hoja2.to_excel(writer, sheet_name='HOJA 2', index=False)

output_file
```
