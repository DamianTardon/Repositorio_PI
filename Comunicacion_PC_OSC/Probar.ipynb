```python
# ============================================================================
# LIBERACIÓN DE RECURSOS
# ============================================================================
# Verificar si hay recursos abiertos.
# Si hay recursos abiertos, obtener el ID del instrumento y cerrar la conexión.

print("\n=== LIBERACIÓN DE RECURSOS ===")
try:    
    instrumento = rm.list_opened_resources()
    
    if instrumento:
        ID = instrumento.query('*IDN?')
        print("ID del instrumento:", ID)
        
        # Cerrar la conexión para liberar recursos.
        instrumento.close()
    else:
        print("No hay recursos abiertos.")
except Exception as e:
    print("Error al conectar con el instrumento:", e)
```
