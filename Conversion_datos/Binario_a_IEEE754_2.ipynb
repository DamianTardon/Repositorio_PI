```python
def half_to_float(word):
    """
    Convierte un valor de 16 bits en formato IEEE754 half-precision a float.
    :param word: Entero de 16 bits que representa el valor half-precision.
    :return: Valor float correspondiente.
    """
    sign = (word >> 15) & 0x01
    exponent = (word >> 10) & 0x1F
    mantissa = word & 0x3FF

    if exponent == 0:
        # Valor subnormal
        value = 2**(-14) * (mantissa / 1024.0)
    elif exponent == 31:
        if mantissa == 0:
            value = float('inf')
        else:
            value = float('nan')
    else:
        value = 2**(exponent - 15) * (1.0 + mantissa / 1024.0)
    
    return -value if sign else value

# Ejemplo de uso con datos recibidos
# data_bytes es una lista de bytes obtenida del puerto serie
# Por ejemplo, con la muestra proporcionada:
data_bytes = [0b00110101,0b10000110]

# Obtener el intervalo de tiempo (little-endian) -- este no funciona correctamente
time_interval_value = data_bytes[0] + (data_bytes[1] << 8)
time_interval_float = half_to_float(time_interval_value)
print(f"Intervalo de tiempo 1: {time_interval_float}")

# Obtener el intervalo de tiempo (big-endian) -- este funciona correctamente
time_interval_value = (data_bytes[0] << 8) + data_bytes[1]
time_interval_float = half_to_float(time_interval_value)
print(f"Intervalo de tiempo 2: {time_interval_float}")

# Prueba para confirmar que el algoritmo funciona bien. Ejemplo:
# 0b0011101111111111 = 0.99951172
time_interval_value = 0b0011101111111111
time_interval_float = half_to_float(time_interval_value)
print(f"Intervalo de tiempo 3: {time_interval_float}")

# Obtener dato de onda (big-endian)
data_bytes = [0b11111111,0b11111111]
wave_value = (data_bytes[0] << 8) + data_bytes[1]
wave_value_float = half_to_float(wave_value)
print(f"Dato Onda 1: {wave_value_float}")
```
