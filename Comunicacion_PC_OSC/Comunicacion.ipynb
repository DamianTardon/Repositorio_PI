```python
# =============================================================================
# Importacion de librerias
# =============================================================================
import pyvisa
```

```python
# =============================================================================
# Busqueda de instrumentos conectados a la PC
# =============================================================================
rm = pyvisa.ResourceManager() # Creacion del administrador de recursos VISA para manejar los instrumentos conectados.
instrumentos = rm.list_resources() # Lista todos los instrumentos conectados a la PC.

print("Instrumentos encontrados:", instrumentos)
```

```python
# =============================================================================
# Verificacion de comunicacion con instrumento
# =============================================================================
try:
    instrumento = rm.open_resource('ASRL4::INSTR') # Abrir el instrumento seleccionado, mediante su direccion.
    
    # Preguntar por la identificacion del instrumento con el comando *IDN?.
    respuesta = instrumento.query('*IDN?')
    print("El instrumento responde:", respuesta)
    instrumento.close() # Cerrar la comunicacion con el instrumento.
except Exception as e:
    print("Error:", e) # Manejo de excepciones en caso de error.

#rm.close() # Cierre del administrador de recursos.
```

```python
try:
    instrumento = rm.open_resource('ASRL4::INSTR')
    
    # Preguntar por la escala de tiempo actual
    escala_tiempo = instrumento.query(':TIMebase:SCALe?')
    print("Escala de tiempo:", escala_tiempo)
    
    instrumento.close()
except Exception as e:
    print("Error:", e)
```

```python
try:
    instrumento = rm.open_resource('ASRL4::INSTR')
    
    # Cambiar escala de tiempo a 1ms/div
    print("Cambiando escala de tiempo a 1ms/div...")
    instrumento.write(':TIMebase:SCALe -ms5') # Analizar que si coloco un valor invalido no lo reconoce pero tampoco lo avisa.
    
    # Verificar que cambió
    nueva_escala = instrumento.query(':TIMebase:SCALe?')
    print("Nueva escala de tiempo:", nueva_escala)
    
    instrumento.close()
except Exception as e:
    print("Error:", e)
```
