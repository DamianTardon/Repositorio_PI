```python
# Valor hexadecimal a convertir.
dato_hexadecimal = "1A"

# Convierte el valor hexadecimal a decimal
# El segundo argumento, 16, le indica a la función int()
# que el número de entrada está en base 16 (hexadecimal).
dato_decimal = int(dato_hexadecimal, 16)

# Imprime el resultado
print(f"El valor hexadecimal '{dato_hexadecimal}' es '{dato_decimal}' en decimal.")
```
