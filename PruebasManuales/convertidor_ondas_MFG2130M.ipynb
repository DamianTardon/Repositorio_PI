```python
import csv

# Configuración de nombres de archivos.
input_filename = 'LI-A1.txt'
output_filename = 'LI-A1.csv'
divisor = 1_000_000
resolucion_14bit = 8191 # Rango de -8191 a 8191.

def process_wave_data():
    values = []

    try:
        # Leer el archivo original (.txt) y cargar los valores en un arreglo.
        with open(input_filename, mode='r', encoding='utf-8') as infile:
            for line in infile:
                clean_line = line.strip()
                if clean_line:
                    val = float(clean_line)
                    values.append(val)

        # Procesamiento de los datos:
        # Primero divide por 1 millón (Voltios).
        # Luego escala a la resolución de 14 bits (enteros).
        processed_values = []
        for v in values:
            valor_voltios = v / divisor

            # Escalado: 1V -> 8191, -1V -> -8191
            valor_escalado = valor_voltios * resolucion_14bit

            # Convertir a entero redondeando al más cercano.
            entero_final = int(round(valor_escalado))

            # Limitar el rango para que no exceda los 14 bits por seguridad.
            if entero_final > 8191: entero_final = 8191
            if entero_final < -8191: entero_final = -8191

            processed_values.append(entero_final)

        # Guardar los nuevos valores en formato entero con coma final.
        with open(output_filename, mode='w', encoding='utf-8') as outfile:
            for v in processed_values:
                outfile.write(f"{v},\n")

        print(f"Proceso completado exitosamente.")
        print(f"Archivo de entrada: {input_filename}")
        print(f"Archivo de salida: {output_filename}")
        print(f"Datos convertidos a 14 bits (rango +/- {resolucion_14bit}).")
        print(f"Se procesaron {len(processed_values)} puntos.")

    except FileNotFoundError:
        print(f"Error: No se encontró el archivo '{input_filename}'.")
    except ValueError:
        print("Error: El archivo .txt contiene datos que no son numéricos.")
    except Exception as e:
        print(f"Ocurrió un error inesperado: {e}")

if __name__ == "__main__":
    process_wave_data()
```
